# MaroFlux Paradigm — Interactive Demo\n\nThis notebook demonstrates the core simulation of the Unified MaroFlux Equation (UME).\n\n**Author:** Marwan Alaa\n**Date:** May 2026

```python
%matplotlib inline\nimport numpy as np\nimport matplotlib.pyplot as plt\nfrom maroflux_solver_complete import MaroFluxSolver
```

## 1. Load Parameters

```python
import yaml\nwith open('parameters/parameter_set.yaml', 'r') as f:\n    params = yaml.safe_load(f)\nprint('Parameters loaded successfully.')
```

## 2. Run Simulation (Short Demo — 5 Days)

```python
params['simulation']['t_end'] = 5.0  # Short demo\nsolver = MaroFluxSolver(params)\nsolver.run()\nprint('Simulation complete.')
```

## 3. Plot Results

```python
t = solver.time_points\nM1 = solver.M1_history\nM2 = solver.M2_history\n\nplt.figure(figsize=(10, 6))\nplt.plot(t, M1, 'r-', label='M1')\nplt.plot(t, M2, 'b-', label='M2')\nplt.xlabel('Time (days)')\nplt.ylabel('Cell Density (cells/cm³)')\nplt.legend()\nplt.title('Macrophage Polarization — MaroFlux')\nplt.grid(True)\nplt.show()
```
